# Paper figure styles and sizing

Use the project kernel after `uv sync`. All values below are synthetic illustration data, not experimental results. No training or experiment records are needed. Project figures automatically use the base style in a local context. Pass a native settings dictionary or a tueplots bundle through `style` to customize the appearance. The venue bundles below are **2024 presets**, not a guarantee about future submission requirements.

```python
import matplotlib.pyplot as plt
from tueplots import bundles

from core import plotting

targets = ["cat", "cat", "dog", "dog", "owl", "owl"]
predictions = ["cat", "dog", "dog", "dog", "cat", "owl"]
```

## Default appearance
The project base style uses a 6×4-inch canvas for general inspection, not a prescribed publication column size. All examples use the same synthetic class-label arrays; these cells do not read saved runs.

```python
fig = plotting.classification(targets, predictions)
plt.show()
```

## Custom style without a venue bundle
A native settings dictionary can change typography, colors, and export DPI. An explicit `figsize` sets the canvas size; the returned axes remain editable. Pass the same style again when saving to apply its export settings. The export line is commented out so this example does not create files.

```python
custom_style = {
    "font.family": "sans-serif",
    "font.size": 10,
    "axes.labelsize": 10,
    "image.cmap": "Blues",
    "savefig.dpi": 200,
}
custom_fig = plotting.classification(
    targets, predictions, style=custom_style, figsize=(4, 3), title=None
)
ax = custom_fig.axes[0]
ax.set_ylabel("Actual class")
ax.set_title(r"Custom $\alpha$", loc="left")
# plotting.save_figure(custom_fig, "experiments/reports/custom-matrix.png", style=custom_style)
plt.show()
```

## NeurIPS 2024
Pass the venue bundle explicitly when drawing. `usetex=False` avoids an external LaTeX installation while retaining MathText support. DejaVu Serif is used for a portable demonstration; use the required installed font for submission. A bundle overrides only the settings it defines; unspecified settings retain their base values.

```python
neurips = bundles.neurips2024(usetex=False) | {"font.serif": ["DejaVu Serif"]}
fig = plotting.classification(targets, predictions, style=neurips)
plt.show()
print("Inches:", neurips["figure.figsize"], "Font size:", neurips["font.size"])
```

## ICML 2024, single column
Change only the native bundle. No project venue registry or additional configuration file is needed.

```python
icml = bundles.icml2024(column="half", usetex=False) | {"font.serif": ["DejaVu Serif"]}
fig = plotting.classification(targets, predictions, style=icml)
plt.show()
print("Inches:", icml["figure.figsize"], "Font size:", icml["font.size"])
```

## Apply to project figures and export
Drawing applies the base style, then explicit `style`, then explicit parameters such as `figsize`, `layout`, or `cmap`. Returned Figure/Axes objects remain editable. Saving also applies local base defaults: pass the same style again to use its export settings. Venue bundles may request tight cropping, changing saved dimensions; set `savefig.bbox` to `None` when exact canvas dimensions are required. Files are written only when you uncomment the export line; existing files require explicit `overwrite=True`.

```python
style = icml | {"savefig.bbox": None}
matrix_fig = plotting.classification(targets, predictions, title=None, style=style)
matrix_fig.axes[0].set_ylabel("True class")
# Pass the same style explicitly when saving.
# plotting.save_figure(matrix_fig, "experiments/reports/icml-example.pdf", style=style)
plt.show()
```
